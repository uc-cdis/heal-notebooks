## About the Study

The study *Transcutaneous Electrical Nerve Stimulation and Pain With Movement in People With Fibromyalgia* was a cluster randomized clinical trial to determine whether transcutaneous electrical nerve stimulation (TENS) is effective for treating pain when combined with physical therapy (PT) in individuals with fibromyalgia. The study was conducted between February 1, 2021 and September 31, 2024, during which time 459 eligible patients were enrolled. Of these, 384 completed baseline data collection and were included in the outcome analysis. The primary outcome was change from baseline to day 60 in movement-evoked pain (scale of 0-10, with 0 indicating no pain and 10 indicating worst pain imaginable) measured during a 5-times sit-and-stand task. In addition, the Patient Global Impression of Change (PGIC), a [HEAL Common Data Element (CDE)](https://www.nih.gov/heal/heal-initiative-requirements/data-sharing-policy/common-data-elements-cdes-program/cdes-repository), was administered.

### Download data using a GUID or manifest

While downloading data through the web can be convenient in some situations, it is often better to download data using the Gen3 command line tool or SDK. This allows you to create a stand-alone, portable notebook or program that can be shared with others, and ensures that your work can be replicated in the future. In addition, in cases where a dataset is very large and/or has many files, downloading through the web quickly becomes inconvenient (if not impossible).

Whenever possible, the HDP indexes datasets at the file level, yielding Globally Unique Identifiers (GUIDs) that can then be used to retrieve the files in perpetuity. Each study's metadata includes a file *manifest*, containing the GUID and other metadata for each of its associated files. Let's retrieve the manifest from this study.

````python
metadata = Gen3Metadata(endpoint='https://healdata.org/')
manifest = metadata.get('HDP01498')['gen3_discovery']['__manifest']
display(Code(json.dumps(manifest, indent=4), language='JSON'))
````

We can use this to download the file using either the Gen3 Python SDK or the `gen3` command (both are included when you install the `gen3` Python package, and are preinstalled in the workspace).
